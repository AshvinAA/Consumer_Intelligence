```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


pd.set_option('display.max_columns', None)

client_df = pd.read_csv('Client.csv')
record_df = pd.read_csv('Record.csv')

df = pd.merge(client_df, record_df, on='Customer_ID', how='inner')
print(f"Merged Data Shape: {df.shape}\n")


print("Top 10 columns with missing values:")
missing_data = df.isnull().sum().sort_values(ascending=False).head(10)
print(missing_data[missing_data > 0])


plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='churn', palette='coolwarm')
plt.title('Company A: Customer Churn Distribution')
plt.xlabel('Churn Status (1 = Left, 0 = Retained)')
plt.ylabel('Number of Customers')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

df.head()
```

```python


# Assuming 'df' is your merged dataset from the very first step.
# If you need to recreate it quickly:
# df = pd.merge(pd.read_csv('Client.csv'), pd.read_csv('Record.csv'), on='Customer_ID', how='inner')

# Set the visual style for executive presentations
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 12})

print("Generating EDA Visualizations for Slide Deck...")

# --- Visual 1: The Cost of Network Failures ---
# Hypothesis: High drop/block rates frustrate customers into leaving.
plt.figure(figsize=(10, 6))
# Create a temporary column to categorize network issues for cleaner visualization
df['Network_Issue_Category'] = pd.qcut(df['drop_blk_Mean'], q=4, labels=['Low', 'Medium', 'High', 'Severe'])

sns.barplot(x='Network_Issue_Category', y='churn', data=df, palette='Reds', errorbar=None)
plt.title('Impact of Network Failures on Churn Rate', fontsize=16, fontweight='bold')
plt.xlabel('Frequency of Dropped/Blocked Calls', fontsize=14)
plt.ylabel('Churn Probability', fontsize=14)
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: '{:.0%}'.format(y))) 
plt.tight_layout()
plt.show()


# --- Visual 2: The "Bill Shock" Effect ---
# Hypothesis: High overage charges cause "bill shock" and drive churn.
plt.figure(figsize=(10, 6))
# Create a total overage feature for visualization
df['Total_Overage'] = df['ovrrev_Mean'] + df['vceovr_Mean'] + df['datovr_Mean']

sns.boxplot(x='churn', y='Total_Overage', data=df, palette='Blues', showfliers=False)
plt.title('Average Overage Charges: Retained vs. Churned', fontsize=16, fontweight='bold')
plt.xlabel('Customer Status (0 = Retained, 1 = Churned)', fontsize=14)
plt.ylabel('Total Overage Revenue ($)', fontsize=14)
plt.tight_layout()
plt.show()


# --- Visual 3: Customer Care Ineffectiveness ---
# Hypothesis: Customers who call support frequently are still churning, meaning support isn't solving the issue.
plt.figure(figsize=(10, 6))
sns.kdeplot(data=df, x='custcare_Mean', hue='churn', common_norm=False, fill=True, palette='coolwarm', clip=(0, 20))
plt.title('Customer Care Call Volume Distribution', fontsize=16, fontweight='bold')
plt.xlabel('Average Monthly Customer Care Calls', fontsize=14)
plt.ylabel('Density', fontsize=14)
plt.tight_layout()
plt.show()
```

## First Hypothesis

Customers are churning due to bill shock of unexpected overage charges and they are frequently abandoning the service without attempting to contact customer care for a solution
